```python
# GPT-2 XL
vocab_size=50257
context_length=1024
num_layers=48
d_model=1600
num_heads=25
d_ff=6400

attention_params = num_layers * (8 * d_model**2 * context_length + 4 * context_length**2 * d_model)
ff_params = num_layers * (6 * d_ff * d_model * context_length)
head_params = 2 * context_length * d_model * vocab_size
total_params = attention_params + ff_params + head_params

print(f"GPT-2 XL")
print(f"Total Parameters: {total_params:.2e}")
print(f"Attention Parameters: {attention_params:.2e} ({attention_params/total_params:.1%})")
print(f"Feedforward Parameters: {ff_params:.2e} ({ff_params/total_params:.1%})")
print(f"LM head Parameters: {head_params:.2e} ({head_params/total_params:.1%})")
```
